# Multiclass -  Neural Network

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.preprocessing import LabelEncoder
```

## DataSet

```python
path = "./dataset-01/Iris/iris.csv"
dataset = pd.read_csv(path)

print(dataset.columns.values)
```

```python
#X = dataset[['SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']]
X = dataset.copy()
X = X.drop(columns = ['Id','Species']) # inplace=True
#X.drop('Id', inplace=True, axis=1)
y = dataset['Species']

print(y[:5])
print(X[:5])
```

```python
y.unique() # Unique values in y (possible classes)
```

```python
class_mapping = {'Iris-setosa': 0, 'Iris-versicolor': 1, 'Iris-virginica': 2}
y = dataset['Species'].map(class_mapping)
print(y)
```

### Splitting in training and testing

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    X, 
    y,
    test_size = 0.3,
    random_state=10
)
print(len(dataset))
print(len(x_train))
print(len(x_test))
```

```python
print(x_train.shape)
print(y_train.shape)
print(y_train[:10])
```

## Tensorflow Model

### Creating the Model

```python
model = Sequential(
    [
        Dense(4, activation='relu'),
        Dense(8, activation='relu'),
        Dense(3, activation='softmax')
    ]
)
```

```python
model.compile(
    loss = tf.keras.losses.SparseCategoricalCrossentropy(),
    optimizer=tf.keras.optimizers.Adam(0.001),
    metrics='accuracy'
)
```

```python
model.fit(x_train,y_train, epochs=100, validation_data=(x_test, y_test))
```

```python
model.summary()
```

### Predictions

```python
test_loss, test_accuracy = model.evaluate(x_test, y_test)
print(test_loss:.4f)
print(test_accuracy)
```

```python
predictions = model.predict(x_test)
print("Predictions = \n", predictions[:5])
```

```python
y_pred = []

for i in range(len(predictions)):
    y_pred.append(np.argmax(predictions[i]))
    print(f"{predictions[i]}, class: {np.argmax(predictions[i])}")
```

```python
acc = np.mean(y_pred == y_test.to_numpy())
print(f"Test set score: {acc:.2f}%")
```
