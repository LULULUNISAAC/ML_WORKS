# Binary Classification - Neural Network

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

## DataSet

```python
path = "dataset-01/iris/"
dataSet = pd.read_csv(path+'iris.csv')
dataSet = dataSet[:100] # Only 2 classes: Iris-setosa and Iris-versicolor
#print(dataSet)
# Shuffling dataframe
data = dataSet.sample(frac=1).reset_index(drop=True)

data[20:40]
```

```python
data["Species"] = data["Species"].apply(lambda x: x.replace("Iris-setosa", "0"))
data["Species"] = data["Species"].apply(lambda x: x.replace("Iris-versicolor", "1"))
data = data.astype(float)
data[:20]
```

```python
X = data.iloc[ : , 1:5 ].values
y = data.iloc[ : , 5:6 ].values
y = y.astype(int)
```

```python
print(X[:5])
print(y[:5])
print(np.shape(X))
print(np.shape(y))

```

# Dataset split

```python
from sklearn.model_selection import train_test_split
import tensorflow as tf
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state=10)
```

```python
print(len(X_test))
```

## Tensorflow Model

### Model

```python
model = Sequential([
    Dense(4, activation = 'relu', name = 'layer1'),
    Dense(1, activation = 'sigmoid', name = 'layers2')
])
```

```python
model.compile(
    loss = tf.keras.losses.BinaryCrossentropy(),
    optimizer = tf.keras.optimizers.Adam(0.01)
)
```

```python
model.fit(
    X_train, y_train,
    epochs= 100
)
```

### Performance Analysis

```python
prediction = model.predict(X_test)
```

```python
binaryPred = (prediction > 0.5).astype(int)

for i in range(len(y_test)):
    print(f"Actual: {y_test[i]} - Prediction: {binaryPred[i]}")
```

```python
accuracy = tf.keras.metrics.CategoricalAccuracy()(y_test, prediction).numpy()
print(f"Accuracy: {accuracy*100:.2f}%")
```
