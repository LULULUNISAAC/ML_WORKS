# Vectorization vs Loop

```python
# Using a Python list
lst = [1, 2, 3, 4, 5]
new_list = [x * 2 for x in lst]
print('List:', new_list)

# Using a NumPy array
arr = np.array([1, 2, 3, 4, 5])  # Now np is defined
new_arr = arr * 2
print('Array:', new_arr)
```

```python
import numpy as np
import time

# Create test data
size = 10**7  # 10 million elements
lst = list(range(size))
arr = np.arange(size)  # Fixed: changed arrange to arange

# Time list operation
start = time.time()
lst_result = [x * 2 for x in lst]
time_lst = time.time() - start

# Time NumPy array operation
start = time.time()
arr_result = arr * 2
time_arr = time.time() - start

print('Time with list:', time_lst)
print('Time with NumPy array:', time_arr)

```
