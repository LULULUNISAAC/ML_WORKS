### Euclidean Distance

```python
import math
import numpy as np
```

```python
def eucDist(v1, v2):
    sm = 0
    for i in range(len(v1)):
        sm += (v1[i] - v2 [i])**2
    return math.sqrt(sm)
```

```python
v1 = [1.2, 2.0, 3.0, 4.0]
v2 = [0.5, 4.5, 5.3, 1.2]
```

```python
eucDist(v1, v2)
```
